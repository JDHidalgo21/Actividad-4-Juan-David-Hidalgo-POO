Ejercicio 6.4: Excepciones

```python
import tkinter as tk
from tkinter import messagebox

class VentanaExcepciones(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Ejercicio 6.4 - Gestión de Excepciones")
        self.geometry("380x300")
        self.resizable(False, False)

        tk.Label(self, text="Demostración Try - Catch - Finally", font=("Arial", 11, "bold")).pack(pady=15)

        tk.Button(self, text="1. Ejecutar Operación Correcta (10 / 2)",
                  command=self.operacion_correcta, width=35, bg="#d4edda").pack(pady=8)

        tk.Button(self, text="2. Ejecutar Operación con Error (10 / 0)",
                  command=self.operacion_error, width=35, bg="#f8d7da").pack(pady=8)

        tk.Label(self, text="Resultado en consola / bitácora:", font=("Arial", 9, "bold")).pack(anchor="w", padx=20, pady=(10, 0))

        self.txt_resultado = tk.Text(self, height=5, width=42)
        self.txt_resultado.pack(padx=20, pady=5)
        self.txt_resultado.insert(tk.END, "Seleccione una opción para ver la ejecución...\n")
        self.txt_resultado.config(state=tk.DISABLED)

    def log(self, mensaje):
        self.txt_resultado.config(state=tk.NORMAL)
        self.txt_resultado.insert(tk.END, mensaje + "\n")
        self.txt_resultado.see(tk.END)
        self.txt_resultado.config(state=tk.DISABLED)

    def operacion_correcta(self):
        self.txt_resultado.delete("1.0", tk.END)
        self.log("--- Inicio del bloque try ---")
        try:
            a = 10
            b = 2
            resultado = a / b
            self.log(f"Operación exitosa: {a} / {b} = {resultado}")
        except ZeroDivisionError:
            self.log("[CATCH] Excepción atrapada: División por cero.")
        finally:
            self.log("[FINALLY] Este bloque se ejecuta siempre.")
        self.log("--- Fin de la ejecución ---")

    def operacion_error(self):
        self.txt_resultado.delete("1.0", tk.END)
        self.log("--- Inicio del bloque try ---")
        try:
            a = 10
            b = 0
            resultado = a / b  # Esto lanza ArithmeticException / ZeroDivisionError
            self.log(f"Resultado: {resultado}")
        except ZeroDivisionError as e:
            self.log(f"[CATCH] Excepción capturada (ArithmeticException): {e}")
        finally:
            self.log("[FINALLY] Este bloque se ejecuta siempre, con o sin error.")
        self.log("--- Fin de la ejecución ---")


if __name__ == "__main__":
    try:
        app = VentanaExcepciones()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```

Ejercicio 6.5: Lanzamiento de exepciones

```python
import tkinter as tk
from tkinter import messagebox

class Vendedor:
    def __init__(self, nombre: str, apellidos: str, edad: int):
        self.verificar_edad(edad)
        self.nombre = nombre
        self.apellidos = apellidos
        self.edad = edad

    def verificar_edad(self, edad: int):
        if edad < 0 or edad > 120:
            raise ValueError("La edad no puede ser negativa ni mayor a 120")
        if edad < 18:
            raise ValueError("El vendedor debe ser mayor de 18 años")

    def imprimir(self) -> str:
        return f"Vendedor: {self.nombre} {self.apellidos}, Edad: {self.edad} años"


class VentanaVendedor(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Ejercicio 6.5 - Lanzamiento de Excepciones")
        self.geometry("360x320")
        self.resizable(False, False)

        tk.Label(self, text="Registro de Vendedor", font=("Arial", 11, "bold")).pack(pady=15)

        # Campos de entrada
        frame_form = tk.Frame(self)
        frame_form.pack(pady=5)

        tk.Label(frame_form, text="Nombre:").grid(row=0, column=0, sticky="e", padx=5, pady=5)
        self.txt_nombre = tk.Entry(frame_form, width=18)
        self.txt_nombre.grid(row=0, column=1, padx=5, pady=5)

        tk.Label(frame_form, text="Apellidos:").grid(row=1, column=0, sticky="e", padx=5, pady=5)
        self.txt_apellidos = tk.Entry(frame_form, width=18)
        self.txt_apellidos.grid(row=1, column=1, padx=5, pady=5)

        tk.Label(frame_form, text="Edad:").grid(row=2, column=0, sticky="e", padx=5, pady=5)
        self.txt_edad = tk.Entry(frame_form, width=18)
        self.txt_edad.grid(row=2, column=1, padx=5, pady=5)

        tk.Button(self, text="Registrar y Verificar Vendedor", command=self.registrar_vendedor, bg="#d1ecf1", width=30).pack(pady=12)

        self.lbl_resultado = tk.Label(self, text="Estado: Esperando datos...", font=("Arial", 9), fg="blue")
        self.lbl_resultado.pack(pady=10)

    def registrar_vendedor(self):
        try:
            nombre = self.txt_nombre.get().strip()
            apellidos = self.txt_apellidos.get().strip()

            if not nombre or not apellidos:
                raise ValueError("Los campos de nombre y apellidos no pueden estar vacíos.")

            edad_str = self.txt_edad.get().strip()
            if not edad_str.isdigit():
                raise ValueError("La edad debe ser un número entero válido.")

            edad = int(edad_str)

            v = Vendedor(nombre, apellidos, edad)

            self.lbl_resultado.config(text=v.imprimir(), fg="green")
            messagebox.showinfo("Éxito", "Vendedor registrado correctamente.")

        except ValueError as e:
            self.lbl_resultado.config(text=f"Error: {e}", fg="red")
            messagebox.showerror("Excepción Capturada", str(e))


if __name__ == "__main__":
    try:
        app = VentanaVendedor()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```

Ejercicio 6.6: Catchs Multiples



```python
import math
import tkinter as tk
from tkinter import messagebox

class CalculosNuméricos:

    @staticmethod
    def calcular_logaritmo(valor: float) -> float:
        if valor <= 0:
            raise ValueError("El valor para el logaritmo neperiano debe ser positivo (> 0).")
        return math.log(valor)

    @staticmethod
    def calcular_raiz_cuadrada(valor: float) -> float:
        if valor <= 0:
            raise ValueError("El valor para la raíz cuadrada debe ser positivo (> 0).")
        return math.sqrt(valor)


class VentanaCalculos(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Ejercicio 6.6 - Cálculos Numéricos y Excepciones")
        self.geometry("380x320")
        self.resizable(False, False)

        tk.Label(self, text="Cálculos Numéricos con Múltiples Excepciones", font=("Arial", 11, "bold")).pack(pady=15)

        # Entrada de datos
        frame_input = tk.Frame(self)
        frame_input.pack(pady=5)

        tk.Label(frame_input, text="Ingrese un valor numérico:").grid(row=0, column=0, padx=5, pady=5, sticky="e")
        self.txt_valor = tk.Entry(frame_input, width=15)
        self.txt_valor.grid(row=0, column=1, padx=5, pady=5)

        # Botones de cálculo
        frame_botones = tk.Frame(self)
        frame_botones.pack(pady=10)

        tk.Button(frame_botones, text="Calcular Logaritmo", command=self.ejecutar_logaritmo, bg="#d1ecf1", width=18).grid(row=0, column=0, padx=5, pady=5)
        tk.Button(frame_botones, text="Calcular Raíz Cuadrada", command=self.ejecutar_raiz, bg="#d4edda", width=18).grid(row=1, column=0, padx=5, pady=5)

        self.lbl_resultado = tk.Label(self, text="Estado: Esperando valor...", font=("Arial", 9), fg="blue", wraplength=340, justify="center")
        self.lbl_resultado.pack(pady=10, padx=20)

    def ejecutar_logaritmo(self):
        try:
            texto = self.txt_valor.get().strip()
            if not texto:
                raise ValueError("El campo de entrada no puede estar vacío.")

            valor = float(texto)
            resultado = CalculosNuméricos.calcular_logaritmo(valor)
            self.lbl_resultado.config(text=f"Logaritmo neperiano de {valor} = {resultado:.4f}", fg="green")

        except ValueError as e:
            mensaje = str(e)
            if "could not convert" in mensaje:
                mensaje_error = "Error de formato (InputMismatchException): Ingrese un valor numérico válido."
            else:
                mensaje_error = f"Excepción aritmética/valor: {mensaje}"

            self.lbl_resultado.config(text=mensaje_error, fg="red")
            messagebox.showerror("Excepción Capturada", mensaje_error)
        except Exception as e:
            self.lbl_resultado.config(text=f"Excepción general: {e}", fg="red")
            messagebox.showerror("Error", str(e))

    def ejecutar_raiz(self):
        try:
            texto = self.txt_valor.get().strip()
            if not texto:
                raise ValueError("El campo de entrada no puede estar vacío.")

            valor = float(texto)
            resultado = CalculosNuméricos.calcular_raiz_cuadrada(valor)
            self.lbl_resultado.config(text=f"Raíz cuadrada de {valor} = {resultado:.4f}", fg="green")

        except ValueError as e:
            mensaje = str(e)
            if "could not convert" in mensaje:
                mensaje_error = "Error de formato (InputMismatchException): Ingrese un valor numérico válido."
            else:
                mensaje_error = f"Excepción aritmética/valor: {mensaje}"

            self.lbl_resultado.config(text=mensaje_error, fg="red")
            messagebox.showerror("Excepción Capturada", mensaje_error)
        except Exception as e:
            self.lbl_resultado.config(text=f"Excepción general: {e}", fg="red")
            messagebox.showerror("Error", str(e))


if __name__ == "__main__":
    try:
        app = VentanaCalculos()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```

6.7 Validación de Campos

```python
import tkinter as tk
from tkinter import messagebox

class Programador:
    def __init__(self, nombre: str, apellidos: str):
        self.validar_campo(nombre, "Nombre")
        self.validar_campo(apellidos, "Apellidos")
        self.nombre = nombre
        self.apellidos = apellidos

    def validar_campo(self, valor: str, etiqueta: str):
        if not valor.strip():
            raise ValueError(f"El campo '{etiqueta}' no puede estar vacío.")
        if len(valor) >= 20:
            raise ValueError(f"El campo '{etiqueta}' no puede tener 20 o más caracteres.")
        for char in valor:
            if char.isdigit():
                raise ValueError(f"El campo '{etiqueta}' no puede contener números (debe ser solo texto).")


class EquipoMaratónProgramación:
    def __init__(self, nombre_equipo: str, universidad: str, lenguaje: str, tamano: int):
        if tamano < 2 or tamano > 3:
            raise ValueError("El tamaño del equipo debe ser de mínimo 2 y máximo 3 integrantes.")
        self.nombre_equipo = nombre_equipo
        self.universidad = universidad
        self.lenguaje = lenguaje
        self.tamano_maximo = tamano
        self.programadores = []

    def esta_completo(self) -> bool:
        return len(self.programadores) >= self.tamano_maximo

    def agregar_programador(self, p: Programador):
        if self.esta_completo():
            raise Exception("Excepción: El equipo ya está lleno (alcanzó el tamaño máximo permitido).")
        self.programadores.append(p)


class VentanaEquipoMaraton(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Ejercicio 6.7 - Validación de Campos en Equipos")
        self.geometry("420x450")
        self.resizable(False, False)

        self.equipo = None

        tk.Label(self, text="Gestión de Equipo de Maratón", font=("Arial", 11, "bold")).pack(pady=10)

        # Marco de Datos del Equipo
        frame_eq = tk.LabelFrame(self, text=" Datos del Equipo ", font=("Arial", 9, "bold"))
        frame_eq.pack(fill="x", padx=15, pady=5)

        tk.Label(frame_eq, text="Nombre Equipo:").grid(row=0, column=0, sticky="e", padx=5, pady=4)
        self.txt_eq_nombre = tk.Entry(frame_eq, width=22)
        self.txt_eq_nombre.grid(row=0, column=1, padx=5, pady=4)

        tk.Label(frame_eq, text="Universidad:").grid(row=1, column=0, sticky="e", padx=5, pady=4)
        self.txt_eq_univ = tk.Entry(frame_eq, width=22)
        self.txt_eq_univ.grid(row=1, column=1, padx=5, pady=4)

        tk.Label(frame_eq, text="Lenguaje:").grid(row=2, column=0, sticky="e", padx=5, pady=4)
        self.txt_eq_lang = tk.Entry(frame_eq, width=22)
        self.txt_eq_lang.grid(row=2, column=1, padx=5, pady=4)

        tk.Label(frame_eq, text="Tamaño (2 o 3):").grid(row=3, column=0, sticky="e", padx=5, pady=4)
        self.txt_eq_tam = tk.Entry(frame_eq, width=22)
        self.txt_eq_tam.grid(row=3, column=1, padx=5, pady=4)

        tk.Button(frame_eq, text="Crear Equipo", command=self.crear_equipo, bg="#d1ecf1").grid(row=4, column=0, columnspan=2, pady=8)

        # Marco de Integrantes
        frame_prog = tk.LabelFrame(self, text=" Agregar Programador ", font=("Arial", 9, "bold"))
        frame_prog.pack(fill="x", padx=15, pady=5)

        tk.Label(frame_prog, text="Nombre:").grid(row=0, column=0, sticky="e", padx=5, pady=4)
        self.txt_prog_nombre = tk.Entry(frame_prog, width=22)
        self.txt_prog_nombre.grid(row=0, column=1, padx=5, pady=4)

        tk.Label(frame_prog, text="Apellidos:").grid(row=1, column=0, sticky="e", padx=5, pady=4)
        self.txt_prog_apellidos = tk.Entry(frame_prog, width=22)
        self.txt_prog_apellidos.grid(row=1, column=1, padx=5, pady=4)

        tk.Button(frame_prog, text="Añadir Integrante", command=self.agregar_integrante, bg="#d4edda").grid(row=2, column=0, columnspan=2, pady=8)

        self.lbl_estado = tk.Label(self, text="Estado: Cree un equipo para empezar.", font=("Arial", 9), fg="blue")
        self.lbl_estado.pack(pady=5)

    def crear_equipo(self):
        try:
            nombre = self.txt_eq_nombre.get().strip()
            univ = self.txt_eq_univ.get().strip()
            lang = self.txt_eq_lang.get().strip()
            tam_str = self.txt_eq_tam.get().strip()

            if not nombre or not univ or not lang or not tam_str:
                raise ValueError("Todos los campos del equipo son obligatorios.")
            if not tam_str.isdigit():
                raise ValueError("El tamaño del equipo debe ser un número entero.")

            tam = int(tam_str)
            self.equipo = EquipoMaratónProgramación(nombre, univ, lang, tam)
            self.lbl_estado.config(text=f"Equipo '{nombre}' creado. Añada integrantes.", fg="green")
            messagebox.showinfo("Éxito", f"Equipo {nombre} creado correctamente.")
        except Exception as e:
            messagebox.showerror("Error de Validación", str(e))
            self.lbl_estado.config(text=f"Error: {e}", fg="red")

    def agregar_integrante(self):
        if not self.equipo:
            messagebox.showwarning("Advertencia", "Primero debe crear y registrar el equipo.")
            return
        try:
            if self.equipo.esta_completo():
                raise Exception("El equipo ya está completo con el máximo de integrantes.")

            nombre = self.txt_prog_nombre.get().strip()
            apellidos = self.txt_prog_apellidos.get().strip()

            prog = Programador(nombre, apellidos)
            self.equipo.agregar_programador(prog)

            total_actual = len(self.equipo.programadores)
            self.lbl_estado.config(text=f"Integrantes añadidos: {total_actual}/{self.equipo.tamano_maximo}", fg="green")
            messagebox.showinfo("Éxito", f"Programador {nombre} {apellidos} añadido con éxito.")

            self.txt_prog_nombre.delete(0, tk.END)
            self.txt_prog_apellidos.delete(0, tk.END)

        except Exception as e:
            messagebox.showerror("Excepción Capturada", str(e))
            self.lbl_estado.config(text=f"Error: {e}", fg="red")


if __name__ == "__main__":
    try:
        app = VentanaEquipoMaraton()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```

Ejercicio 6.8: Lectura de Archivos

```python
import os
import tkinter as tk
from tkinter import filedialog, messagebox

class GestorArchivo:

    @staticmethod
    def crear_archivo_prueba(nombre_archivo="prueba.txt"):
        if not os.path.exists(nombre_archivo):
            with open(nombre_archivo, "w", encoding="utf-8") as f:
                f.write("Línea 1: Bienvenidos a la lectura de archivos en POO.\n")
                f.write("Línea 2: Este archivo demuestra el manejo de flujos de datos.\n")
                f.write("Línea 3: Universidad Nacional de Colombia - Actividad 4.\n")

    @staticmethod
    def leer_archivo(ruta_archivo: str) -> str:
        if not os.path.exists(ruta_archivo):
            raise FileNotFoundError(f"El archivo '{ruta_archivo}' no existe en el sistema.")

        contenido = ""
        # Simulando el flujo de lectura de bytes y caracteres
        with open(ruta_archivo, "r", encoding="utf-8") as archivo:
            for linea in archivo:
                contenido += linea
        return contenido


class VentanaLeerArchivo(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Ejercicio 6.8 - Lectura de Archivos")
        self.geometry("450x380")
        self.resizable(False, False)

        tk.Label(self, text="Lector de Archivos de Texto", font=("Arial", 11, "bold")).pack(pady=10)

        # Botones de acción superior
        frame_btns = tk.Frame(self)
        frame_btns.pack(pady=5)

        tk.Button(frame_btns, text="Crear/Restaurar 'prueba.txt'", command=self.generar_prueba, bg="#d1ecf1", width=22).grid(row=0, column=0, padx=5)
        tk.Button(frame_btns, text="Examinar Archivo...", command=self.buscar_archivo, bg="#d4edda", width=18).grid(row=0, column=1, padx=5)

        tk.Label(self, text="Contenido del archivo:", font=("Arial", 9, "bold")).pack(anchor="w", padx=20, pady=(10, 0))

        self.txt_contenido = tk.Text(self, height=10, width=50)
        self.txt_contenido.pack(padx=20, pady=5)
        self.txt_contenido.insert(tk.END, "Haga clic en 'Crear/Restaurar' o 'Examinar' para leer un archivo...\n")
        self.txt_contenido.config(state=tk.DISABLED)

        self.lbl_estado = tk.Label(self, text="Estado: Listo", font=("Arial", 9), fg="blue")
        self.lbl_estado.pack(pady=5)

    def generar_prueba(self):
        try:
            GestorArchivo.crear_archivo_prueba("prueba.txt")
            self.cargar_y_mostrar("prueba.txt")
            messagebox.showinfo("Éxito", "Archivo 'prueba.txt' creado y leído correctamente.")
        except Exception as e:
            messagebox.showerror("Error", str(e))

    def buscar_archivo(self):
        ruta = filedialog.askopenfilename(title="Seleccionar archivo de texto", filetypes=[("Archivos de texto", "*.txt"), ("Todos los archivos", "*.*")])
        if ruta:
            self.cargar_y_mostrar(ruta)

    def cargar_y_mostrar(self, ruta):
        try:
            texto = GestorArchivo.leer_archivo(ruta)
            self.txt_contenido.config(state=tk.NORMAL)
            self.txt_contenido.delete("1.0", tk.END)
            self.txt_contenido.insert(tk.END, texto)
            self.txt_contenido.config(state=tk.DISABLED)
            self.lbl_estado.config(text=f"Archivo leído con éxito: {os.path.basename(ruta)}", fg="green")
        except Exception as e:
            self.lbl_estado.config(text=f"Excepción de E/S: {e}", fg="red")
            messagebox.showerror("Excepción de E/S", str(e))


if __name__ == "__main__":
    try:
        app = VentanaLeerArchivo()
        app.mainloop()
    except Exception as e:
        print(f"Entorno sin interfaz gráfica o error de ejecución: {e}")
```
